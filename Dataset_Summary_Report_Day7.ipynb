## 8. Observations

- The dataset contains **25 rows and 7 columns**, covering student ID, name, age, course, marks, attendance percentage, and city.
- Available features are a mix of **numerical** (`Student_ID`, `Age`, `Marks`, `Attendance_Percentage`) and **categorical** (`Name`, `Course`, `City`) data.

- Multiple students share the same `City` and `Course` values, indicating some natural grouping potential for further analysis (e.g. by course or city) beyond this summary report.
